count-based vectorization

Bag of words

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
corpus=['the cat sit on the mat','the dog sit on the mat','the dog run']
```

```python
cv=CountVectorizer(stop_words='english') # the tipli sozleri silir . tekrarlanir deye onemli goturmesin deye
x=cv.fit_transform(corpus)
```

```python
cv.get_feature_names_out()
```

```python
x.toarray()
```

TF_IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tf=TfidfVectorizer()
```

```python
tf.fit_transform(corpus)
```

```python
tf.get_feature_names_out()
```

```python
tf.idf_
```

##WORD2VEC

```python
!pip install gensim
```

```python
sentences=[
    ['the','cat','sits','on','the','mat'],
    ['the','dog','sits','on','the','mat'],
    ['the','car','drives','on','the','road'],
    ]
```

```python
from gensim.models import Word2Vec
```

```python
model=Word2Vec(sentences,
               min_count=1,
               vector_size=10,
               window=2,
               sg=1, #sg=1 skip gram sg=0 cbow
               )
```

```python
model.wv['cat']
```

#Glove-Global Vectors

```python
import gensim.downloader as api
glove=api.load('glove-wiki-gigaword-50')
```

```python
glove.most_similar('cat',topn=3)
```

```python
print(glove.most_similar(positive=['cat','meal'],
                         negative=['mouse'],topn=1))
```

#Fast Text

```python
from gensim.models import FastText

sentence=[
    'the cat sit on the mat'.split(),
    'the dog sit on the mat'.split(),
    'i love my cat'.split(),
]
```

```python
model=FastText(
    sentence,
    vector_size=50,
    window=3,
    min_count=1,
    epochs=100,
)
```

```python
model.wv['cats']
```

```python
model.wv.most_similar('cats',topn=2)
```
